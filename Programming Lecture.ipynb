**Steps involved:**2. **Padding and FFT**: Zero-phase shifts the window into an `N=512` array buffer and applies the FFT.

4. **Envelope Interpolation & Phase Randomization**: Upsamples the smoothed envelope back to its original size (`mY`), generates random phases (`pY`), and maps magnitude and phase to form half of the complex spectrum.

## Harmonic plus Stochastic Model (Single Frame)

5. **Harmonic Time-Domain Resynthesis**: Transforms the harmonic spectrum (`Yh`) back to the time-domain using an Inverse FFT (`yh`).
6. **Stochastic Resynthesis**: Applies random uniform phase to the smoothed stochastic envelope to construct a complex conjugate symmetrical spectrum, and performs an Inverse FFT to generate the stochastic time-domain signal (`yst`).

```python
(fs,x1)=UF.wavread('/home/devang-bordoloi/ASP/sms-tools-materials-main/sounds/rain.wav')
(fs,x2)=UF.wavread('/home/devang-bordoloi/ASP/sms-tools-materials-main/sounds/soprano-E4.wav')
M=N=512
w=get_window('hann',M)
x1w=x1[10000:10000+M]*w
x2w=x2[10000:10000+M]*w
mX1,pX1=DFT.dftAnal(x1w,w,N)
mX2,pX2=DFT.dftAnal(x2w,w,N)
smoothf=.2
mX2smooth1=resample(np.maximum(-200,mX2),int(mX2.size*smoothf))
mX2smooth2=resample(mX2smooth1,mX1.size)
balancef=.7
mY=balancef*mX2smooth2+(1-balancef)*mX1
y=DFT.dftSynth(mY,pX1,N)
fig1=plt.figure()
plt.plot(mX1)
plt.plot(mX2)
plt.plot(mX2smooth2)
plt.plot(mY)
fig2=plt.figure()
plt.plot(y)
```

## Sinusoidal Model Analysis and Synthesis

This script applies the full Sinusoidal Model to analyze and resynthesize an entire audio file (`piano.wav`). The sound is decomposed into a set of sinusoidal tracks and then reconstructed.

**Steps involved:**
1. **Sinusoidal Analysis**: Iterates over the signal frame by frame to detect spectral peaks and track them over time, generating trajectories of frequency, magnitude, and phase for each sine track (`SM.sineModelAnal`).
2. **Phase-less Synthesis**: Synthesizes the time-domain signal (`y`) solely from the frequency and magnitude trajectories, assuming zero initial phase and generating phase continuously (`SM.sineModelSynth` without phase).
3. **Phased Synthesis**: Synthesizes another version (`y1`) utilizing the explicitly tracked phase information along with frequency and magnitude trajectories.
4. **Output Generation**: Writes the phase-less synthesized audio out as a separate wave file (`test2.wav`).
5. **Visualization**: Plots the synthesized time-domain signals overlapped against the original audio waveform to compare the results of both phase-less and phased synthesis.

```python
inputfile = '/home/devang-bordoloi/ASP/sms-tools-materials-main/sounds/piano.wav'
window='hamming'
M=1001
N=2048
t=-100
minSineDur=0.01
maxnSines=150
freqDevOffset=30
freqDevSlope=0.002
Ns=512
H=128
(fs,x)=UF.wavread(inputfile)
w=get_window(window,M)
tfreq, tmag, tphase = SM.sineModelAnal(x, fs, w, N, H, t, maxnSines, minSineDur, freqDevOffset, freqDevSlope)
y=SM.sineModelSynth(tfreq, tmag, np.array([]), Ns,H,fs)
y1=SM.sineModelSynth(tfreq, tmag, tphase, Ns,H,fs)
UF.wavwrite(y,fs,'/home/devang-bordoloi/ASP/test2.wav')
fig1=plt.figure()
plt.plot(x)
plt.plot(y)
fig2=plt.figure()
plt.plot(x)
plt.plot(y1)
plt.show()
```
